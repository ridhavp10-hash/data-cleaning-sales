```python
import pandas as pd
df = pd.read_csv("train.csv")
print(df.shape)
df.isna().sum()
```

```python
# Flag before filling, so nothing is hidden
df["age_missing"] = df["Age"].isna()
df["has_cabin"] = df["Cabin"].notna()

# Fill Age with the median for each title group
df["Title"] = df["Name"].str.extract(r", (\w+)\.")
df["Age"] = df["Age"].fillna(df.groupby("Title")["Age"].transform("median"))

# Embarked: fill with the most common value
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Cabin: too empty to fill, so keep only the deck letter
df["Deck"] = df["Cabin"].str[0].fillna("Unknown")
df = df.drop(columns=["Cabin"])

df.isna().sum()
```

```python
df.to_csv("titanic_clean.csv", index=False)
```

```python
df["Name"] = df["Name"].str.strip()
df["Sex"] = df["Sex"].str.title()
df.isna().sum()
```

```python
df.to_csv("titanic_clean.csv", index=False)
```
